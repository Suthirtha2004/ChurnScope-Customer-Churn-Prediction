```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
```

Data Cleaning and Encoding

```python
df = pd.read_csv("/content/customer_data.csv")
df.head()
df.info()

df['TotalCharges'] = pd.to_numeric(df['TotalCharges'],errors='coerce')
print(df['TotalCharges'].isna().sum())
print(df['TotalCharges'].dtypes)

df['TotalCharges'] = df['TotalCharges'].fillna(df['TotalCharges'].median())
print(df['TotalCharges'].isna().sum())

```

```python
df['Churn_Encoded'] = df['Churn'].map({
    "No" : 0,
    "Yes" : 1
})

df.replace("No phone service","No",inplace=True)
df.replace("No internet service","No",inplace=True)

df.head(5)
```

```python
# Label Encoding and One-Hot Encoding

cat_features = ["Partner","Dependents","PhoneService","MultipleLines","OnlineSecurity","OnlineBackup","DeviceProtection","TechSupport","StreamingTV",
                "StreamingMovies","PaperlessBilling"]

for col in cat_features:
  print(col,df[col].unique())
```

```python
def label_encode_values(df,cat_features):
  for i in cat_features:

    df[i] = df[i].map({
        "Yes":1,
        "No":0
    })
  return df

df = label_encode_values(df,cat_features)
df['gender'] = df['gender'].map({
    "Female":1,
    "Male":0
})

print(df.head())
```

```python
# One Hot encoding
from sklearn.preprocessing import OneHotEncoder

categ_features = ['InternetService','Contract','PaymentMethod']
encoder = OneHotEncoder(sparse_output=False)
encoded_data = encoder.fit_transform(df[categ_features])

encoded_df = pd.DataFrame(
    encoded_data,
    columns = encoder.get_feature_names_out(categ_features),
    index = df.index
)

final_df = pd.concat(
    [df.drop(columns=categ_features),encoded_df],
    axis=1
)

print("\nOne-Hot Encoded Data")
print(final_df.head())
print(final_df.columns.tolist())
```

Train-test-split and MinMax Scaling

```python
from sklearn.preprocessing import MinMaxScaler
from sklearn.model_selection import train_test_split

y = final_df['Churn_Encoded']
x = final_df.drop(columns=['customerID','Churn','Churn_Encoded'])

print(x.columns)

x_train , x_test , y_train , y_test = train_test_split(
    x,y,
    test_size = 0.2,
    random_state = 42,
    stratify = y
)
```

```python
# MinMax Scaling

num_features = ['TotalCharges','MonthlyCharges','tenure']
scaler = MinMaxScaler()

x_train = x_train.copy()
x_test = x_test.copy()

x_train[num_features] = scaler.fit_transform(
    x_train[num_features]
)

x_test[num_features] = scaler.transform(
    x_test[num_features]
)

print(x_train[num_features].describe())
print(x_test[num_features].describe())
```

Decision Tree Classifier - Gini Index and Entropy Criterion

```python
from sklearn.metrics import accuracy_score, classification_report , confusion_matrix
from sklearn.tree import DecisionTreeClassifier , plot_tree
import seaborn as sns

clf_gini = DecisionTreeClassifier(criterion ="gini",random_state=100,max_depth=3,min_samples_leaf=5)
clf_gini.fit(x_train,y_train)
y_pred = clf_gini.predict(x_test)

print(f"Classification Report :\n {classification_report(y_pred,y_test)}")
print(f"Accuracy Score : {accuracy_score(y_pred,y_test)*100:.2f}")
print(f"Confusion Matrix \n: {confusion_matrix(y_pred,y_test)}")

cmf = confusion_matrix(y_pred,y_test)
plt.figure(figsize=(8,5))
sns.heatmap(cmf,annot=True,fmt='g',cmap='Blues',cbar=True,
            xticklabels=['Predicted No Churn','Predicted Churn'],
            yticklabels=['Actual No Churn','Actual Churn'])

plt.title('Confusion Matrix Heatmap')
plt.xlabel('Predicted Labels')
plt.ylabel('True Labels')
plt.show()
```

```python
# Plot Tree

plt.figure(figsize=(30,10))

plot_tree(
    clf_gini,
    feature_names = list(x.columns),
    class_names = ["No Churn","Churn"],
    filled = True,
    rounded = True
)

plt.show()
```

```python
clf_entropy = DecisionTreeClassifier(criterion="entropy",max_depth=3,min_samples_leaf=5,random_state=100)
clf_entropy.fit(x_train,y_train)
y_pred = clf_entropy.predict(x_test)

print(f"Classification Report :\n {classification_report(y_pred,y_test)}")
print(f"Accuracy Score : {accuracy_score(y_pred,y_test)*100:.2f}")
print(f"Confusion Matrix \n: {confusion_matrix(y_pred,y_test)}")

cmf = confusion_matrix(y_pred,y_test)
plt.figure(figsize=(8,5))
sns.heatmap(cmf,annot=True,fmt='g',cmap='Blues',cbar=True,
            xticklabels=['Predicted No Churn','Predicted Churn'],
            yticklabels=['Actual No Churn','Actual Churn'])

plt.title('Confusion Matrix Heatmap')
plt.xlabel('Predicted Labels')
plt.ylabel('True Labels')
plt.show()
```

```python
plt.figure(figsize=(30,10))

plot_tree(
    clf_entropy,
    feature_names = list(x.columns),
    class_names = ["No Churn","Churn"],
    filled = True,
    rounded = True
)

plt.show()
```
